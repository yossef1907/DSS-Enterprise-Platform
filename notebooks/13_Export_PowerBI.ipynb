```python
import pandas as pd
import sqlite3
import os

db_path = r'C:\Users\user\DSS_Project\database\dss_project.db'
conn = sqlite3.connect(db_path)

export_path = r'C:\Users\user\DSS_Project\exports\powerbi'
os.makedirs(export_path, exist_ok=True)

print("✅ Connected to database")
print(f"✅ Export folder ready: {export_path}")
```

```python
eco = pd.read_sql_query("SELECT * FROM ecommerce", conn)
eco['date'] = pd.to_datetime(eco['date'])
eco['year']    = eco['date'].dt.year
eco['month']   = eco['date'].dt.month
eco['quarter'] = eco['date'].dt.quarter

eco.to_csv(f'{export_path}\\ecommerce_powerbi.csv', index=False)
print(f"✅ ecommerce_powerbi.csv → {len(eco):,} rows")
```

```python
segments = pd.read_sql_query("SELECT * FROM customer_segments", conn)
segments.to_csv(f'{export_path}\\customer_segments_powerbi.csv', index=False)
print(f"✅ customer_segments_powerbi.csv → {len(segments):,} rows")
```

```python
churn = pd.read_sql_query("SELECT * FROM telco_churn", conn)
churn.to_csv(f'{export_path}\\churn_powerbi.csv', index=False)
print(f"✅ churn_powerbi.csv → {len(churn):,} rows")
```

```python
roi = pd.read_sql_query("SELECT * FROM roi_by_channel", conn)
roi.to_csv(f'{export_path}\\roi_powerbi.csv', index=False)
print(f"✅ roi_powerbi.csv → {len(roi):,} rows")

roi_region = pd.read_sql_query("SELECT * FROM roi_by_region", conn)
roi_region.to_csv(f'{export_path}\\roi_region_powerbi.csv', index=False)
print(f"✅ roi_region_powerbi.csv → {len(roi_region):,} rows")
```

```python
forecast = pd.read_sql_query("SELECT * FROM sales_forecast", conn)
forecast.to_csv(f'{export_path}\\forecast_powerbi.csv', index=False)
print(f"✅ forecast_powerbi.csv → {len(forecast):,} rows")
```

```python
recs = pd.read_sql_query("SELECT * FROM recommendations", conn)
recs.to_csv(f'{export_path}\\recommendations_powerbi.csv', index=False)
print(f"✅ recommendations_powerbi.csv → {len(recs):,} rows")
```

```python
rules = pd.read_sql_query("SELECT * FROM association_rules", conn)
rules.to_csv(f'{export_path}\\basket_rules_powerbi.csv', index=False)
print(f"✅ basket_rules_powerbi.csv → {len(rules):,} rows")
```

```python
eco2 = pd.read_sql_query("SELECT * FROM ecommerce", conn)
eco2['date']  = pd.to_datetime(eco2['date'])
eco2['month'] = eco2['date'].dt.to_period('M').astype(str)

monthly_kpi = eco2.groupby('month').agg(
    total_revenue = ('total_amount', 'sum'),
    total_orders  = ('order_id', 'count'),
    avg_order_val = ('total_amount', 'mean'),
    unique_customers = ('customer_id', 'nunique')
).reset_index()

monthly_kpi['total_revenue'] = monthly_kpi['total_revenue'].round(2)
monthly_kpi['avg_order_val'] = monthly_kpi['avg_order_val'].round(2)

monthly_kpi.to_csv(f'{export_path}\\monthly_kpi_powerbi.csv', index=False)
print(f"✅ monthly_kpi_powerbi.csv → {len(monthly_kpi):,} rows")
```

```python
conn.close()

print("\n📂 Files ready for Power BI:")
print("="*50)
files = os.listdir(export_path)
for f in sorted(files):
    size = os.path.getsize(f'{export_path}\\{f}')
    print(f"   ✅ {f:45} ({size/1024:.1f} KB)")

print(f"\n✅ Total: {len(files)} files ready")
print(f"📁 Location: {export_path}")
```
