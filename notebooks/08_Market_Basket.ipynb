```python
import pandas as pd

# Load datasets
eco = pd.read_csv(r'C:\Users\user\DSS_Project\data\clean\ecommerce_clean.csv')
mkt = pd.read_csv(r'C:\Users\user\DSS_Project\data\clean\marketing_clean.csv')
ret = pd.read_csv(r'C:\Users\user\DSS_Project\data\clean\retail_clean.csv',
                  nrows=100000)  # sample عشان الـ memory

# E-Commerce Revenue
eco_revenue = eco['total_amount'].sum()
eco_orders  = eco['order_id'].nunique()
eco_aov     = eco_revenue / eco_orders

# Marketing Revenue
mkt_revenue = mkt['sales_revenue_usd'].sum()
mkt_budget  = mkt['marketing_budget_usd'].sum()
mkt_roi     = (mkt_revenue - mkt_budget) / mkt_budget * 100

# Retail Revenue
ret_revenue = ret['total_sales'].sum()

print("=" * 50)
print("💰 TOTAL REVENUE SUMMARY")
print("=" * 50)
print(f"\nE-Commerce:")
print(f"   Total Revenue : ${eco_revenue:,.2f}")
print(f"   Total Orders  : {eco_orders:,}")
print(f"   AOV           : ${eco_aov:,.2f}")

print(f"\nMarketing:")
print(f"   Revenue       : ${mkt_revenue:,.2f}")
print(f"   Budget        : ${mkt_budget:,.2f}")
print(f"   ROI           : {mkt_roi:.1f}%")

print(f"\nRetail (sample):")
print(f"   Revenue       : ${ret_revenue:,.2f}")

print(f"\nCombined Total:")
print(f"   Total Revenue : ${eco_revenue + mkt_revenue:,.2f}")
print("=" * 50)
```

```python
mkt = pd.read_csv(r'C:\Users\user\DSS_Project\data\clean\marketing_clean.csv')

# ROI per channel
roi_channel = mkt.groupby('sales_channel').agg(
    budget  = ('marketing_budget_usd', 'sum'),
    revenue = ('sales_revenue_usd', 'sum')
).reset_index()

roi_channel['roi'] = (
    (roi_channel['revenue'] - roi_channel['budget']) / 
     roi_channel['budget'] * 100
).round(2)

print("📊 ROI by Channel:")
print(roi_channel.sort_values('roi', ascending=False).to_string(index=False))

# ROI per region
roi_region = mkt.groupby('region').agg(
    budget  = ('marketing_budget_usd', 'sum'),
    revenue = ('sales_revenue_usd', 'sum')
).reset_index()

roi_region['roi'] = (
    (roi_region['revenue'] - roi_region['budget']) / 
     roi_region['budget'] * 100
).round(2)

print("\n📊 ROI by Region:")
print(roi_region.sort_values('roi', ascending=False).to_string(index=False))

# ROI per season
roi_season = mkt.groupby('season').agg(
    budget  = ('marketing_budget_usd', 'sum'),
    revenue = ('sales_revenue_usd', 'sum')
).reset_index()

roi_season['roi'] = (
    (roi_season['revenue'] - roi_season['budget']) / 
     roi_season['budget'] * 100
).round(2)

print("\n📊 ROI by Season:")
print(roi_season.sort_values('roi', ascending=False).to_string(index=False))

# ROI per segment
roi_segment = mkt.groupby('customer_segment').agg(
    budget  = ('marketing_budget_usd', 'sum'),
    revenue = ('sales_revenue_usd', 'sum')
).reset_index()

roi_segment['roi'] = (
    (roi_segment['revenue'] - roi_segment['budget']) / 
     roi_segment['budget'] * 100
).round(2)

print("\n📊 ROI by Segment:")
print(roi_segment.sort_values('roi', ascending=False).to_string(index=False))
```

```python
# الميزانية الحالية
total_budget = mkt['marketing_budget_usd'].sum()
print(f"Total Budget: ${total_budget:,.2f}")

# استراتيجية توزيع الميزانية المحسّنة
optimized_allocation = {
    # Channels — نركز على Wholesale وOnline
    'Wholesale':    0.35,  # زيادة من 19.6% لـ 35%
    'Online':       0.30,  # ثابت تقريباً
    'Social Media': 0.20,  # زيادة
    'Retail Store': 0.10,  # تقليل
    'Direct Sales': 0.05,  # تقليل كبير
}

# Segments — نركز على Corporate
optimized_segments = {
    'Corporate': 0.45,  # زيادة كبيرة — الوحيد إيجابي
    'VIP':       0.30,  # زيادة
    'Regular':   0.20,  # تقليل
    'New':       0.05,  # تقليل كبير
}

# Seasons — نركز على Q4 وQ3
optimized_seasons = {
    'Q4': 0.40,  # زيادة كبيرة
    'Q3': 0.30,  # زيادة
    'Q2': 0.20,  # تقليل
    'Q1': 0.10,  # تقليل كبير
}

# حساب الـ ROI المتوقع بعد التحسين
# بناءً على أحسن channel + segment + season

best_combo = mkt[
    (mkt['sales_channel'] == 'Wholesale') &
    (mkt['customer_segment'] == 'Corporate') &
    (mkt['season'] == 'Q4')
]

if len(best_combo) > 0:
    best_roi = (
        (best_combo['sales_revenue_usd'].sum() - 
         best_combo['marketing_budget_usd'].sum()) /
         best_combo['marketing_budget_usd'].sum() * 100
    )
    print(f"\n✅ Best Combination ROI:")
    print(f"   Wholesale + Corporate + Q4")
    print(f"   ROI = {best_roi:.2f}%")

# توقع الـ ROI بعد إعادة التوزيع
current_roi = -41.1
improvement = (
    # Channel improvement
    ((-32.15 - (-41.1)) * 0.35) +
    ((-38.60 - (-41.1)) * 0.30) +
    # Segment improvement  
    ((8.03 - (-41.1)) * 0.45) +
    # Season improvement
    ((-25.91 - (-41.1)) * 0.40)
) / 3

optimized_roi = current_roi + improvement

print(f"\n📊 ROI Analysis:")
print(f"   Current ROI     : {current_roi:.1f}%")
print(f"   Expected improvement: +{improvement:.1f}%")
print(f"   Optimized ROI   : {optimized_roi:.1f}%")

# حساب Revenue المتوقع
current_revenue = mkt['sales_revenue_usd'].sum()
optimized_revenue = total_budget * (1 + optimized_roi/100)
revenue_increase = optimized_revenue - current_revenue

print(f"\n💰 Revenue Impact:")
print(f"   Current Revenue  : ${current_revenue:,.2f}")
print(f"   Optimized Revenue: ${optimized_revenue:,.2f}")
print(f"   Revenue Increase : ${revenue_increase:,.2f}")
```

```python
# نحسب كل combinations ونلاقي الأفضل
results = []

channels = mkt['sales_channel'].unique()
segments = mkt['customer_segment'].unique()
seasons  = mkt['season'].unique()
regions  = mkt['region'].unique()

for channel in channels:
    for segment in segments:
        for season in seasons:
            for region in regions:
                combo = mkt[
                    (mkt['sales_channel'] == channel) &
                    (mkt['customer_segment'] == segment) &
                    (mkt['season'] == season) &
                    (mkt['region'] == region)
                ]
                
                if len(combo) >= 5:
                    budget  = combo['marketing_budget_usd'].sum()
                    revenue = combo['sales_revenue_usd'].sum()
                    roi     = (revenue - budget) / budget * 100
                    
                    results.append({
                        'channel':  channel,
                        'segment':  segment,
                        'season':   season,
                        'region':   region,
                        'budget':   round(budget, 2),
                        'revenue':  round(revenue, 2),
                        'roi':      round(roi, 2)
                    })

combos_df = pd.DataFrame(results)
combos_df = combos_df.sort_values('roi', ascending=False)

print("🏆 Top 20 Best Combinations:")
print(combos_df.head(20).to_string(index=False))

print(f"\n✅ Best ROI: {combos_df['roi'].max():.2f}%")
print(f"✅ Combinations with Positive ROI: {(combos_df['roi'] > 0).sum()}")
print(f"✅ Combinations with ROI > 50%: {(combos_df['roi'] > 50).sum()}")
print(f"✅ Combinations with ROI > 100%: {(combos_df['roi'] > 100).sum()}")
```

```python
# Save top combinations
combos_df.to_csv(
    r'C:\Users\user\DSS_Project\exports\best_marketing_combinations.csv', 
    index=False)

combos_df.to_csv(
    r'C:\Users\user\DSS_Project\exports\powerbi\best_combinations_powerbi.csv',
    index=False)

# Top 5 strategy
top5 = combos_df.head(5)

print("🏆 OPTIMAL MARKETING STRATEGY")
print("="*60)
print(f"Total Positive ROI Combinations : 126")
print(f"Best ROI Achieved               : 106.47%")
print(f"Average ROI Top 15              : {combos_df.head(15)['roi'].mean():.2f}%")

print(f"\nTop 5 Recommended Combinations:")
print(top5[['channel','segment','season','region','roi']].to_string(index=False))

# Budget reallocation
positive_combos = combos_df[combos_df['roi'] > 0]

focused_revenue = positive_combos['revenue'].sum()
focused_budget  = positive_combos['budget'].sum()
focused_roi     = (focused_revenue - focused_budget) / focused_budget * 100
current_revenue = mkt['sales_revenue_usd'].sum()
revenue_gain    = focused_revenue - current_revenue

print(f"\nIf we focus budget on positive ROI combinations:")
print(f"   Focused Budget  : ${focused_budget:,.2f}")
print(f"   Focused Revenue : ${focused_revenue:,.2f}")
print(f"   Focused ROI     : {focused_roi:.2f}%")
print(f"   Revenue Gain    : ${revenue_gain:,.2f}")

print("\nSaved to exports successfully")
```

```python
# Smart Budget Reallocation
total_budget = mkt['marketing_budget_usd'].sum()

# الاستراتيجية الجديدة — نوزع الـ Budget على أحسن combinations
top_combos = combos_df[combos_df['roi'] > 40].copy()

# نحسب نسبة كل combination من الـ Budget
top_combos['budget_weight'] = (
    top_combos['roi'] / top_combos['roi'].sum()
)

# نوزع الـ Budget الكامل على الـ top combinations
top_combos['allocated_budget'] = (
    top_combos['budget_weight'] * total_budget
).round(2)

# نحسب الـ Revenue المتوقع بنفس الـ ROI
top_combos['expected_revenue'] = (
    top_combos['allocated_budget'] * 
    (1 + top_combos['roi'] / 100)
).round(2)

total_expected_revenue = top_combos['expected_revenue'].sum()
total_expected_roi = (
    (total_expected_revenue - total_budget) / 
     total_budget * 100
)

print("="*60)
print("SMART BUDGET REALLOCATION STRATEGY")
print("="*60)
print(f"\nTotal Budget Available  : ${total_budget:,.2f}")
print(f"Current Revenue         : ${mkt['sales_revenue_usd'].sum():,.2f}")
print(f"Current ROI             : -41.1%")
print(f"\nAfter Reallocation:")
print(f"Expected Revenue        : ${total_expected_revenue:,.2f}")
print(f"Expected ROI            : {total_expected_roi:.2f}%")
print(f"Revenue Improvement     : ${total_expected_revenue - mkt['sales_revenue_usd'].sum():,.2f}")

print(f"\nTop Combinations Budget Allocation:")
print(top_combos[['channel','segment','season','region',
                   'roi','allocated_budget',
                   'expected_revenue']].head(10).to_string(index=False))

# Save
top_combos.to_csv(
    r'C:\Users\user\DSS_Project\exports\smart_reallocation.csv',
    index=False)

top_combos.to_csv(
    r'C:\Users\user\DSS_Project\exports\powerbi\smart_reallocation_powerbi.csv',
    index=False)

print("\nSaved successfully")
```
